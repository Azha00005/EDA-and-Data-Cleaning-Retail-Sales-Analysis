**Очищення Даних:**

Визначте та обробіть відсутні дані (видалення або заміна).

```python
df = pd.DataFrame(file)
```

```python
print(df.isnull().sum())
```

```python
df_dropped = df.dropna()
print(df_dropped)
```

Виявлення та видалення аномалій, які можуть викривити аналіз,
за допомогою інтерквартильного діапазону або Z-оцінок.

```python
Q1 = df['Sales'].quantile(0.25) # 25-й перцентиль
Q3 = df['Sales'].quantile(0.75) # 75-й перцентиль
IQR = Q3 - Q1

# Визначаємо межі (стандартна формула: 1.5 * IQR)
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Нижня межа: {lower_bound}")
print(f"Верхня межа: {upper_bound}")
```

Забезпечте узгодженість категорійних даних (наприклад, єдине
написання категорій).

```python
text_cols = ['Ship Mode', 'Segment', 'Country', 'City', 'State', 'Region', 'Category', 'Sub-Category']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
print(df.head())
```

**Дослідницький Аналіз Даних (EDA)**

Проведіть EDA для виявлення патернів, трендів і кореляцій між
продажами, регіонами та категоріями товарів.

```python
# Групуємо дані, щоб побачити зв'язок між Регіоном та Категорією
pattern_table = df.groupby(['Region', 'Category'])['Sales'].sum().unstack()
print(pattern_table)

numeric_df = df.select_dtypes(include=['number'])
print("Матриця кореляції числових змінних")
print(numeric_df.corr()['Sales'])
```

Візуалізуйте тренди за допомогою лінійних графіків, стовпчастих
діаграм та гістограм (Matplotlib/Seaborn).

```python
plt.figure(figsize=(10, 6))
sns.heatmap(pattern_table, annot=True, fmt=",.0f", cmap="YlGnBu", linewidths=.5)
plt.title('Матриця продажів: Як регіон впливає на категорію товару?')
plt.xlabel('Категорія товару')
plt.ylabel('Регіон')
plt.show()
```

Аналізуйте сезонні тренди продажів і виявляйте найкращі
регіони/товари

```python
file['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')

file['Month'] = file['Order Date'].dt.month_name()

#АНАЛІЗ 1: СЕЗОННІСТЬ (Коли купують?)
seasonal_sales = file.groupby('Month')['Sales'].sum()
seasonal_sales.plot(kind='bar', title='Продажі по місяцях', figsize=(10, 5))
plt.show()

#АНАЛІЗ 2: РЕГІОНИ (Де купують?)
region_sales = file.groupby('Region')['Sales'].sum()

# Малюємо кругову діаграму (бублик)
region_sales.plot(kind='pie', autopct='%1.0f%%', title='Частка продажів по регіонах')
plt.show()

#АНАЛІЗ 3: ТОП ТОВАРИ (Що купують?)
# Сортуємо від найбільшого до найменшого і беремо перші 5
top_products = file.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)

print("ТОП-5 ТОВАРІВ ЗА ВИРУЧКОЮ")
print(top_products)
```

```python
#Скользящее среднее
monthly_trend = file.set_index('Order Date').resample('ME')['Sales'].sum()

# Рахуємо середнє за 6 місяців (згладжуємо графік)
rolling_mean = monthly_trend.rolling(window=6).mean()

plt.figure(figsize=(12, 6))
plt.plot(monthly_trend, label='Реальні продажі', color='blue', alpha=0.3)
plt.plot(rolling_mean, label='Тренд (Скользящее)', color='red', linewidth=3)

plt.title('Аналіз тренду: Куди рухаються продажі?')
plt.legend()
plt.grid(True)
plt.show()
```
